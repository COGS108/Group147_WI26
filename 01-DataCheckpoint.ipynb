#### 1. Header 
CO2 Emissions in U.S created per state from 1970 to 2021

#### 2. Description

##### Variables
year: The year for data entry

state-name: Geographical location

sector-name: Economic sector 

fuel-name: Fuel type (coal, petroleum, natural gas, and all fuels combined)

value: The amount of emissions in million metric tons of CO2.

The dataset contains annual estimates of CO₂ emissions for each U.S. state from 1970 through 2021. Emissions are disaggregated by state, economic sector, and fuel type, along with the corresponding quantity of CO₂ produced. The economic sectors include industrial, residential, commercial, transportation, and electric power. This sectoral breakdown enables analysis of how emissions sources vary across states and over time. This provides us with plenty of information about CO2 emissions per state, specifically from where the emisions come from and help us find data for the counties that we want to work with.

Every row contains year, state name, one sector that creates CO2 emissions, type of fuel that is used and how much carbon emissions have been created. The listing of about all the states in the U.S. it makes it easier to compare between states or counties when needed. The time frame that this dataset has provides a lot of flexibility to work with a specific period of time, whether is 1 or more years. And the separation the sectors that create CO2 emissions into different categories allows us to identify which sectors are related to A.I. datacenters and which sectors are not. Lastly, the measurements used to estimate the CO2 emissions values per state are "metric tons CO2"(a measure of greenhouse gas mass equal to 1000 kilograms of CO2), one metric ton of CO₂ is roughly equivalent to the emissions from driving a gasoline car about 2,500–2,700 miles. Using this measurement simplifies the analysis by avoiding extremely large numbers with many zeros, making the data easier to interpret and compare.
