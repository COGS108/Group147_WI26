```python
# loading clean datasets
import pandas as pd
df_dc = pd.read_csv("data/02-processed/df_datacenters_clean.csv")
df_countygdp = pd.read_csv("data/02-processed/df_countygdp_clean.csv")
df_emissions = pd.read_csv("data/02-processed/df_emissions_clean.csv")
df_water = pd.read_csv("data/02-processed/df_water_clean.csv")
df_electricity = pd.read_csv("data/02-processed/df_electricity_clean.csv")
```

```python
# df emissions
df_co2 = df_emissions[(df_emissions['sector-name']=='Industrial carbon dioxide emissions') & (df_emissions['year']==2021) & (df_emissions['fuel-name'] == 'All Fuels')]
df_co2 = df_co2.rename(columns={
    'state-name': 'state',
})

# new df counties: containing all counties 
df_counties = df_countygdp[['county']].copy()

#saving df counties to data/02processed
df_counties.to_csv('data/02-processed/df_counties.csv',index=False)

#aggregating total sqft per county in df_dc
total_sqft = df_dc.groupby(['state','county'])['sqft'].sum().reset_index()
total_sqft.rename(columns={'sqft':'total_sqft'}, inplace=True)

# getting columns in df electricity for 2021
elec_2021_cols = [col for col in df_electricity.columns if "2021" in col]
df_electricity['elec_price_2021'] = df_electricity[elec_2021_cols].mean(axis=1)
df_electricity = df_electricity[['state', 'elec_price_2021']].copy()
```

```python
# cleaning names of counties and states functions

def clean_county(county):
    county = str(county).strip().lower()

    for suffix in [" county", 
               " parish", 
               " borough", 
               " census area", 
               " municipality", 
               "city and borough"]:
        if county.endswith(suffix):
            county = county[:-len(suffix)]
    return county

def clean_state(state):
    state = str(state).strip().lower()
    return state

for df in [df_dc, df_countygdp, total_sqft, df_water]:
    df['county'] = df['county'].apply(clean_county)
    df['state'] = df['state'].apply(clean_state)

df_co2['state'] = df_co2['state'].apply(clean_state)
df_electricity['state'] = df_electricity['state'].apply(clean_state)
```

#### Merging datasets

```python
# merging county gdp with total sqft from df_dc
df = df_countygdp.merge(total_sqft, on=['state','county'],how='left')

#filling nan sqft with 0
df['total_sqft'] = df['total_sqft'].fillna(0)

#merging df w df_water
df = df.merge(df_water, on=['state','county'], how='left')

#merging df w df_co2
df = df.merge(df_co2, on=['state'], how='left')

#merging df w df_electricity

df = df.merge(df_electricity, on=['state'], how='left')

#dropping irrelevant columns 

df
```

#### Creating treatment variables

```python

# has_dc: 1 if county has a data center, 0 otherwise
df['has_dc'] = df['total_sqft'].apply(lambda x: 1 if x > 0 else 0)

# large_dc if sqft over 1,000,000, 1 if yes, 0 otherwise
df['large_dc'] = df['total_sqft'].apply(lambda x: 1 if x > 1_000_000 else 0)
```

#### Section 1 of EDA - Economic Variable Analysis

This part compares the baseline GDP for counties, as well as visualizing economic variables through SMD and boxplots.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

econ_vars = [
    "gdp2022",
    "gdp2023",
    "gdp2024",
    "elec_price_2021",
    "total_sqft"
]

df_econ = df[econ_vars + ["has_dc", "large_dc"]].copy()

print(df_econ.describe())
```

```python
plt.figure(figsize=(6,5))
sns.boxplot(x="has_dc", y="gdp2022", data=df)
plt.title("Baseline GDP (2022) by Data Center Presence")
plt.xlabel("Has Data Center (0=No, 1=Yes)")
plt.ylabel("GDP 2022 Growth (%)")
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.boxplot(data=df[econ_vars])
plt.xticks(rotation=45)
plt.title("Distribution of Economic Variables")
plt.show()
```

```python
gdp_means = df.groupby("has_dc")["gdp2022"].mean()
print("Mean GDP 2022 by Data Center Presence:")
print(gdp_means)

def compute_smd(data, treatment, variable):
    group1 = data[data[treatment] == 1][variable]
    group0 = data[data[treatment] == 0][variable]
    
    mean1 = group1.mean()
    mean0 = group0.mean()
    
    sd_pooled = np.sqrt((group1.var() + group0.var()) / 2)
    
    return (mean1 - mean0) / sd_pooled

smd_results = {}

for var in econ_vars:
    smd_results[var] = compute_smd(df, "has_dc", var)

smd_df = pd.DataFrame.from_dict(smd_results, orient="index", columns=["SMD"])
print(smd_df)

plt.figure(figsize=(7,5))

plt.axvline(x=0.1, linestyle="--")
plt.axvline(x=-0.1, linestyle="--")

sns.barplot(x=smd_df["SMD"], y=smd_df.index)

plt.title("Standardized Mean Differences (Data Center vs No Data Center)")
plt.xlabel("SMD")
plt.show()
```

#### Section 2 of EDA - Environmental Data Analysis 



This part compates the state-level CO2 emmisons (per capita), water usage (indistrial per capita), and electricity prices. It produced visualizatins to better understand trends and differences between states. 

Notes:
- Data comes from multiple years; some datasets may not align exactly.
- Missing values have been handled by dropping incomplete rows.
- Regression models are illustrative and not predictive becasue at the moment is from inconsistent years. 

```python
import statsmodels.formula.api as smf
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Create derived columns
# CO2 per capita (if TP-TotPop exists)
if 'TP-TotPop' in df.columns:
    df['co2_per_capita'] = df['value'] / df['TP-TotPop']
else:
    df['co2_per_capita'] = df['value']  # fallback if no population column

# CO2 per GDP
df['co2_per_gdp'] = df['value'] / df['gdp2022'] 

# Water withdrawals per capita (using TO-Wtotl)
df['water_per_capita'] = df['TO-Wtotl'] / df['TP-TotPop'] if 'TP-TotPop' in df.columns else df['TO-Wtotl']


# 2. State-level electricity prices
state_elec = df.groupby('state')['elec_price_2021'].mean().reset_index()

plt.figure(figsize=(12,5))
sns.barplot(data=state_elec, x='state', y='elec_price_2021')
plt.xticks(rotation=45, fontsize=8)
plt.title("State-level Average Electricity Price (2021)")
plt.ylabel("Electricity Price ($/kWh)")
plt.show()

# Regression model for CO2 per capita

# Uses has_dc, gdp2022, total water, and state
model_co2 = smf.ols(
    'co2_per_capita ~ has_dc + gdp2022 + Q("TO-Wtotl") + C(state)',
    data=df
).fit()

print(model_co2.summary())

# Visualizations

# CO2 per capita by state
plt.figure(figsize=(12,5))
sns.barplot(data=df, x='state', y='co2_per_capita')
plt.xticks(rotation=45, fontsize=8)
plt.title("CO2 Emissions per Capita by State")
plt.ylabel("CO2 per Capita")
plt.show()

# Water per capita by state
plt.figure(figsize=(12,5))
sns.barplot(data=df, x='state', y='water_per_capita')
plt.xticks(rotation=45, fontsize=8)
plt.title("Water Withdrawals per Capita by State")
plt.ylabel("Water per Capita (units of TO-Wtotl)")
plt.show()

# CO2 vs Water scatterplot
plt.figure(figsize=(7,5))
sns.scatterplot(data=df, x='water_per_capita', y='co2_per_capita', hue='state')
plt.title("CO2 per Capita vs Water per Capita")
plt.xlabel("Water per Capita")
plt.ylabel("CO2 per Capita")
plt.legend(bbox_to_anchor=(1.05, 1), loc=2)
plt.show()
```

# Section 3 of EDA- Regression Analysis 

In this section, we run simple OLS (ordinary least squares) regression models using has_dc (a binary indicator for whether a county hosts at least one data center) as the main explanatory variable. We focus on economic outcomes using county GDP (2022-2024) and environmental outcomes using CO2 emissions and electricity price. Because CO2 and electricity price measures are available at the state level in the datasets we currently have, we interpret environmental regressions as broarder association and avoid making strong county-level claims. This regression analysis aligns with oour hypothesis that host locations may experience both economic gains and environmental costs

```python
#Prepare a dataset for regression 
import statsmodels.formula.api as smf 
```

```python
# list of the variables needed for the regression models
# Columns we need for the regressions in this section
# NOTE: CO2 is currently stored in the column named "value"
needed_cols = ["has_dc", "gdp2022", "gdp2023", "gdp2024", "value", "elec_price_2021"]

# Check if any needed columns are missing
missing = [col for col in needed_cols if col not in df.columns]

if len(missing) > 0:
    print("Missing columns in df:", missing)
    print("Fix: Run the earlier merge/cleaning cells that create df, then run this section again.")
else:
    # Drop only rows missing the variables we need
    df_reg = df.dropna(subset=needed_cols).copy()

    # Rename CO2 column so regressions are easier to read
    df_reg = df_reg.rename(columns={"value": "co2"})

    print("Rows in original df:", len(df))
    print("Rows used for regression (df_reg):", len(df_reg))
    print("Key columns available:", [c for c in ["has_dc", "gdp2022", "gdp2023", "gdp2024", "co2", "elec_price_2021"] if c in df_reg.columns])

```

### ECONOMIC MODELS (county-level outcomes)
Outcome: GDP (county level), predictor: has_dc, control: gdp2022 baseline GDP, and using robust standard errors (HC3)



```python
# GDP 2023 regression
model_gdp2023 = smf.ols(
    formula="gdp2023 ~ has_dc + gdp2022",  # regression formula
    data=df_reg                            # regression dataset
).fit(cov_type="HC3")                      # robust standard errors

print("GDP 2023 Regression Results")
print(model_gdp2023.summary())


# GDP 2024 regression
model_gdp2024 = smf.ols(
    formula="gdp2024 ~ has_dc + gdp2022",
    data=df_reg
).fit(cov_type="HC3")                      # robust standard errors

print("\nGDP 2024 Regression Results")
print(model_gdp2024.summary())
```

### Environmental Models 
CO2 and electricity price are state-level variables


```python
# CO2 regression
model_co2 = smf.ols(
    formula="co2 ~ has_dc",
    data=df_reg
).fit(cov_type="HC3")

print("\nCO2 Emissions Regression Results")
print(model_co2.summary())


# Electricity price regression
model_elec = smf.ols(
    formula="elec_price_2021 ~ has_dc",
    data=df_reg
).fit(cov_type="HC3")

print("\nElectricity Price Regression Results")
print(model_elec.summary())
```

### Regression Interpretation
The regression analysis finds limited evidence that counties hosting data centers experience significantly different economic outcomes compared counties wihtout data centers once baseline GDP levels are considered. In both 2023 and 2024 models, the coefficient for has_dc is not statistically significant, suggesting that the presence of data centers alone doesn't strongly predict county GDP differences in this dataset. Environemntal models also show no statistically significant relationship between data centers presence and CO2 emissions, probably since the emissions are measired at the state level instead of the county level. The electricity price model show a small positive association with data center presence.
