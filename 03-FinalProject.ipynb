## Research Question

How do U.S. counties hosting large-scale data centers differ from comparable non-host counties in terms of economic outcomes (GDP) and environmental impacts (CO₂ emissions, water and electricity use)?


## Background and Prior Work

Technology has been invented in order to make human lives easier, but as this technology keeps evolving, the world starts changing too. With the invention of artificial intelligence (AI), many controversies have come up when it comes to the benefits and disadvantages that it brings to the environment, the economy, education, and our everyday life. Some studies suggest that AI can support environmental sustainability by improving efficiency, energy systems and allow better monitoring of environmental conditions[1](#cite_ref-1) [2](##cite_ref-2). At the same time, the fast expansion of AI has created new forms of resource demand that raise sustainability concerns. As AI data centers increase across economic sections, understanding both its potential benefits and environmental consequences has become very important. 
 
AI systems rely on very large computer infrastructure that is mainly located in data centers in order to support tasks such as cloud computing, machine learning training, and large data processing. Reviews of data center operations indicate that energy intensive servers and cooling systems make these facilities among the fastest growing sources of industrial electricity consumption worldwide. [3](#cite_ref-3) In the United States, the growth of AI has contributed to a fast expansion of data centers infrastructure, increasing overall demand for electricity and placing tension on power systems.[4](#cite_ref-4) As AI continues to grow, the physical infrastructure needed to support these technologies has become a main factor in looking at sustainability.

Studies demonstrate that data centers contribute to a significant and growing environmental burden in the AI era. Guidi et al, find that U.S. data centers consume a major portion of national electricity and that AI data centers are only increasing this. Large chunks of data centers can put tension on regional power grids, increase peak demand, and sometimes prolong dependency on carbon intensive energy sources.[5](#cite_ref-5) In addition to energy consumption, water usage for cooling is a major concern, especially in dry regions. Although solutions like advanced cooling systems, waste heat reuse, and water positive designs have been proposed, their adoption remains inconsistent across facilities.[6](#cite_ref-6) Together, these studies show that AI data centers generate major environmental impacts related to electricity demand, greenhouse emissions, and water consumption. 

In addition to environmental concerns, data centers are often promoted as an economic growth impact in host communities. Research indicates that regulatory environments can increase data center valuation, and attract private investment, encouraging companies to locate large facilities in certain regions.[7](#cite_ref-7) Data centers may also contribute to local tax revenue infrastructure development and employment opportunities, particularly in technology and engineering sections.[8](#cite_ref-8) However, community research suggests that these economic benefits are often going along with local costs, including increased demand for electricity and water, land use pressure, and environmental stress.[9](#cite_ref-9) While prior work studies have examined economic outcomes or environmental impacts separately, only a few directly compare these topics at the county level across the United States. The lack of research motivates a modified evaluation of how U.S. counties hosting large-scale data centers differ from comparable non-hosting counties in terms of environmental impacts (CO2 emissions, electricity use, and water consumption) and economic outcomes (employment growth and municipal tax revenue).

```python
# Data center size distribution
plt.figure(figsize=(7,5))
sns.histplot(df["total_sqft"], bins=40)
plt.title("Distribution of Data Center Square Footage")
plt.xlabel("Total Data Center Square Footage")
plt.ylabel("Count")
plt.show()
```

```python
# Log-scaled version 
plt.figure(figsize=(7,5))
sns.histplot(np.log1p(df["total_sqft"]), bins=40)
plt.title("Log Distribution of Data Center Square Footage")
plt.xlabel("Log(Total Sqft + 1)")
plt.ylabel("Count")
plt.show()
```

```python
def plot_variable_relationship(data, variable, title):

    plt.figure(figsize=(6,5))
    sns.boxplot(data=data, x="has_dc", y=variable)

    plt.title(title)
    plt.xlabel("Data Center Presence (0 = No, 1 = Yes)")
    plt.ylabel(variable)

    plt.show()

    print("\nMean values:")
    print(data.groupby("has_dc")[variable].mean())
```

```python
# GDP Analysis
print("\n----- GDP Analysis -----")

plot_variable_relationship(
    df,
    "gdp2022",
    "Baseline GDP Growth (2022) by Data Center Presence"
)

plot_variable_relationship(
    df,
    "gdp2023",
    "GDP Growth (2023) by Data Center Presence"
)

plot_variable_relationship(
    df,
    "gdp2024",
    "GDP Growth (2024) by Data Center Presence"
)


# GDP Trend visualizations
gdp_cols = ["gdp2022","gdp2023","gdp2024"]

gdp_means = df.groupby("has_dc")[gdp_cols].mean().T

plt.figure(figsize=(7,5))

plt.plot(gdp_means.index, gdp_means[0], marker="o", label="No Data Center")
plt.plot(gdp_means.index, gdp_means[1], marker="o", label="Has Data Center")

plt.title("Average GDP Growth Trend by Data Center Presence")
plt.xlabel("Year")
plt.ylabel("Average GDP Growth (%)")
plt.legend()

plt.show()
```

```python
# Electricity Price Analysis

plot_variable_relationship(
    df,
    "elec_price_2021",
    "Electricity Price by Data Center Presence"
)

plt.figure(figsize=(6,5))

sns.scatterplot(
    data=df,
    x="elec_price_2021",
    y="total_sqft",
    hue="has_dc"
)

plt.title("Electricity Price vs Data Center Size")
plt.xlabel("Electricity Price")
plt.ylabel("Total Data Center Square Footage")

plt.show()
```

```python
#Gets rid of the top 1% of the outliers for data can be fair (explain below)
df['host_county'] = (df['total_sqft'] > 0).astype(int)
```

```python
#Cretes visualization does the numbers
county_counts = df['host_county'].value_counts().rename({
    0: "No Data Center",
    1: "Has Data Center"
})

print(county_counts)

plt.figure(figsize=(6,4))

sns.countplot(data=df, x='host_county')

plt.xticks([0,1], ['No Data Center','Has Data Center'])
plt.title("Number of Counties With vs Without Data Centers")
plt.xlabel("County Type")
plt.ylabel("Number of Counties")

plt.show()
```

# Section 3 of EDA- Regression Analysis 

In this section, we run simple OLS (ordinary least squares) regression models using has_dc (a binary indicator for whether a county hosts at least one data center) as the main explanatory variable. We focus on economic outcomes using county GDP (2022-2024) and environmental outcomes using CO2 emissions and electricity price. Because CO2 and electricity price measures are available at the state level in the datasets we currently have, we interpret environmental regressions as broarder association and avoid making strong county-level claims. This regression analysis aligns with oour hypothesis that host locations may experience both economic gains and environmental costs

```python
#Prepare a dataset for regression 
import statsmodels.formula.api as smf 
```

```python
# list of the variables needed for the regression models
# Columns we need for the regressions in this section
# NOTE: CO2 is currently stored in the column named "value"
needed_cols = ["has_dc", "gdp2022", "gdp2023", "gdp2024", "value", "elec_price_2021"]

# Check if any needed columns are missing
missing = [col for col in needed_cols if col not in df.columns]

if len(missing) > 0:
    print("Missing columns in df:", missing)
    print("Fix: Run the earlier merge/cleaning cells that create df, then run this section again.")
else:
    # Drop only rows missing the variables we need
    df_reg = df.dropna(subset=needed_cols).copy()

    # Rename CO2 column so regressions are easier to read
    df_reg = df_reg.rename(columns={"value": "co2"})

    print("Rows in original df:", len(df))
    print("Rows used for regression (df_reg):", len(df_reg))
    print("Key columns available:", [c for c in ["has_dc", "gdp2022", "gdp2023", "gdp2024", "co2", "elec_price_2021"] if c in df_reg.columns])
```

### ECONOMIC MODELS (county-level outcomes)
Outcome: GDP (county level), predictor: has_dc, control: gdp2022 baseline GDP, and using robust standard errors (HC3)

```python
# GDP 2023 regression
model_gdp2023 = smf.ols(                   # Create a regression model using statsmodels
    formula="gdp2023 ~ has_dc + gdp2022",  # regression formula
    data=df_reg                            # regression dataset
).fit(cov_type="HC3")                      # robust standard errors

print("GDP 2023 Regression Results")       #prints title
print(model_gdp2023.summary())             #prints full regression result table


# GDP 2024 regression
model_gdp2024 = smf.ols(                   # Create a regression model using statsmodels 
    formula="gdp2024 ~ has_dc + gdp2022",  #regression formula
    data=df_reg                            # Cleaned dataset
).fit(cov_type="HC3")                      # robust standard errors

print("\nGDP 2024 Regression Results")       # Prints title
print(model_gdp2024.summary())               # Prints full regression table 
```

These regression models describe whether the counties hosting data centers experience different GDP outputs while controlling for baseline GDP in 2022. In both the 2023 and 2024 models, the coefficient of  has_dc is not statistically significant, indicating that counties with data centers don’t show significantly different GDP levels compared to counties without data centers. But, the coefficient for gdp2022 is statistically significant in both models, which suggest that previous economic performance is a strong predictor of future GDP. In general, these results indicate that baseline economic conditions explain GDP variation more strongly than the presence of data centers alone. 

### Environmental Models 
CO2 and electricity price are state-level variables

```python
# CO2 regression
model_co2 = smf.ols( # Create a regression model using statsmodels 
    formula="co2 ~ has_dc",  # Regression formula, outcome variable = co2, predictor variable = has_dc
    data=df_reg     # This is to use the cleaned regression dataframe
).fit(cov_type="HC3")    # This fits the regression model and helps to make the results more reliable

print("\nCO2 Emissions Regression Results")    # Prints a clear title
print(model_co2.summary())                      # Prints the full regression results table


# Electricity price regression
model_elec = smf.ols(                     # Creates another regression model
    formula="elec_price_2021 ~ has_dc",    # regression formula, outcome variable = elec_price_2021, predictor variable = has_dc
    data=df_reg                             # Used for a cleaned regression dataframe
).fit(cov_type="HC3")                         # Run the regression model 

print("\nElectricity Price Regression Results")  # Prints the full regression results table
print(model_elec.summary())
```

These environmental modes examine whether the presence of data centers is associated with differences in CO2 emissions and electricity prices. The CO2 regression shows that the coefficient for has_dc is not statistically significant, which shows that the counties hosting data centers don’t appear to have significantly different CO2 emission levels compared to those without data centers. An explanation for this is that the CO2 emissions levels are measured at the state level, which may not capture county level environmental impacts. For the electricity price model, it shows a small positive association between data center presence and electricity prices, but the result is marginally significant. This suggests that increased electricity demand from data centers may slightly influence local electricity pricing but the evidence is weak to confirm a clear relationship. 

#### Regression Interpretation

Overall, the regression results provide limited evidence that the presence of data centers significantly affects county level economic outcomes. While baseline GDP strongly predicts future GDP levels, the presence of data centers alone doesn’t appear to have a significant difference in GDP. Environmental models also show no statistically significant relationship between data center presence and CO2 emissions, likely due to emission being measured at the state level rather than the county level. The electricity price model suggest a small positive association with data center presence, which shows that increased electricity demand may slightly influence local electricity markets.
